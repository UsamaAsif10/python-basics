Suppose we have two stacks and no other temporary variable. Construct a queue data structure using only these two stacks. Use the following template for `Q1Queue` when developing the solution and raise an `IndexError` when dequeue is called on an empty queue. Next, create a instance `q1` and check if the enqueue and dequeue operations are correctly working.

````python
class Q1Queue:
    def __init__(self):
        self.stack1 = []  # Stack to handle enqueues
        self.stack2 = []  # Stack to handle dequeues

    def enqueue(self, item):
        """Add an item to the queue."""
        self.stack1.append(item)

    def dequeue(self):
        """Remove and return the front item from the queue."""
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())

        if not self.stack2:
            raise IndexError("dequeue from an empty queue")

        return self.stack2.pop()

    def is_empty(self):
        """Check if the queue is empty."""
        return not self.stack1 and not self.stack2

    def __str__(self):
        """Return a string representation of the queue."""
        if self.stack2:
            return "Queue: " + str(self.stack2[::-1] + self.stack1)
        else:
            return "Queue: " + str(self.stack1)

q1 = Q1Queue()

# Enqueue elements
q1.enqueue(10)
q1.enqueue(20)
q1.enqueue(30)

try:
    print("Dequeued:", q1.dequeue())  # Should print 10
    print("Dequeued:", q1.dequeue())  # Should print 20
    print("Dequeued:", q1.dequeue())  # Should print 30
    print("Dequeued:", q1.dequeue())  # Should raise an IndexError
except IndexError as e:
    print("Error:", e)

print("Is queue empty?", q1.is_empty())

q1.enqueue(40)
q1.enqueue(50)
print("Dequeued:", q1.dequeue())  # Should print 40

print(q1)
````

#### Question 2&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;(10 Points)
Given two queues with their standard operations (e.g., enqueue, dequeue, size), implement a stack with its standard operations (pop, push, top, size). Next, create a instance `q2` and push 1, 2, 3, 4 and then pop once.

````python
from collections import deque

class StackUsingQueues:
    def __init__(self):
        self.queue1 = deque()  # The active queue for current elements
        self.queue2 = deque()  # The helper queue for manipulation

    def push(self, item):
        """Add an item to the stack."""
        self.queue1.append(item)

    def pop(self):
        """Remove and return the top item from the stack."""
        if self.is_empty():
            raise IndexError("pop from an empty stack")

        while len(self.queue1) > 1:
            self.queue2.append(self.queue1.popleft())

        top_item = self.queue1.popleft()

        self.queue1, self.queue2 = self.queue2, self.queue1

        return top_item

    def top(self):
        """Return the top item from the stack without removing it."""
        if self.is_empty():
            raise IndexError("top from an empty stack")

        while len(self.queue1) > 1:
            self.queue2.append(self.queue1.popleft())

        top_item = self.queue1.popleft()
        self.queue2.append(top_item)  # Put it back into queue2

        self.queue1, self.queue2 = self.queue2, self.queue1

        return top_item

    def is_empty(self):
        """Check if the stack is empty."""
        return len(self.queue1) == 0

    def size(self):
        """Return the number of items in the stack."""
        return len(self.queue1)

    def __str__(self):
        """Return a string representation of the stack."""
        return "Stack: " + str(list(self.queue1)[::-1])

q2 = StackUsingQueues()

q2.push(1)
q2.push(2)
q2.push(3)
q2.push(4)

popped_item = q2.pop()

print("Popped item:", popped_item)  # Should print 4
print(q2)  # Should show Stack: [3, 2, 1]
````

Write a Python program to reverse a string using a stack. Use the following template when developing the solution. Next, create a instance `q3` and check if the reverse operation is correctly working. You should raise an `IndexError` when pop is called on an empty stack.

````python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        """Add an item to the stack."""
        self.items.append(item)

    def pop(self):
        """Remove and return the top item from the stack."""
        if self.is_empty():
            raise IndexError("pop from an empty stack")
        return self.items.pop()

    def is_empty(self):
        """Check if the stack is empty."""
        return len(self.items) == 0

    def size(self):
        """Return the number of items in the stack."""
        return len(self.items)

    def __str__(self):
        """Return a string representation of the stack."""
        return "Stack: " + str(self.items)

def reverse_string(input_string):
    """Reverse a string using a stack."""
    stack = Stack()

    for char in input_string:
        stack.push(char)

    reversed_string = ""
    while not stack.is_empty():
        reversed_string += stack.pop()

    return reversed_string

q3 = "hello, world!"

reversed_q3 = reverse_string(q3)

print("Original string:", q3)
print("Reversed string:", reversed_q3)  # Should print "!dlrow ,olleh"
````

````python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        """Add an item to the stack."""
        self.items.append(item)

    def pop(self):
        """Remove and return the top item from the stack."""
        if self.is_empty():
            raise IndexError("pop from an empty stack")
        return self.items.pop()

    def peek(self):
        """Return the top item from the stack without removing it."""
        if self.is_empty():
            raise IndexError("peek from an empty stack")
        return self.items[-1]

    def is_empty(self):
        """Check if the stack is empty."""
        return len(self.items) == 0

    def size(self):
        """Return the number of items in the stack."""
        return len(self.items)

    def __str__(self):
        """Return a string representation of the stack."""
        return "Stack: " + str(self.items)


def sort_stack(original_stack):
    """Sorts a stack in ascending order using a temporary stack."""
    temp_stack = Stack()

    while not original_stack.is_empty():
        current = original_stack.pop()

    
        while not temp_stack.is_empty() and temp_stack.peek() > current:
            original_stack.push(temp_stack.pop())

        temp_stack.push(current)

    while not temp_stack.is_empty():
        original_stack.push(temp_stack.pop())

    return original_stack

q4 = Stack()
q4.push(3)
q4.push(5)
q4.push(1)
q4.push(4)
q4.push(2)
q4.push(8)

sorted_q4 = sort_stack(q4)
print(sorted_q4)  # Should print Stack: [1, 2, 3, 4, 5, 8]
````

```python
    ({[]})()
```
is balanced, while:
```python
    ({[}])
```
is not.


````python
class Stack:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if self.is_empty():
            raise IndexError("pop from an empty stack")
        return self.items.pop()

    def peek(self):
        if self.is_empty():
            raise IndexError("peek from an empty stack")
        return self.items[-1]

    def size(self):
        return len(self.items)

def q5_parentheses(s):
    stack = Stack()
    matching_parentheses = {')': '(', '}': '{', ']': '['}
    for char in s:
        if char in '({[':
            stack.push(char)
        elif char in ')}]':
            if stack.is_empty() or stack.pop() != matching_parentheses[char]:
                return False
    return stack.is_empty()
example1 = "({[]})()"
example2 = "({[}])"

print("Example1 is balanced:", q5_parentheses(example1))  
print("Example2 is balanced:", q5_parentheses(example2)) 
````

#### Question 6&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;(10 Points)
Write a Python program to check if an expression has duplicate parenthesis or not using stacks. Use the method `q6_duplicate_parentheses`, which returns `True` if it has duplicates. Otherwise, `False`.

For example: 
```python
    (((x+(y))+(m+n)))
```
has duplicate parenthesis since the whole expression is surrounded by two pairs of brackets. While
```python
    ((x+y)+(m+n))
```
does not have any duplicate parenthesis since no subsexpression is surrounded by duplicate brackets.

````python
def q5_parentheses(input_string):
    """Check if the parentheses in the input string are balanced."""
    # Initialize an empty stack
    stack = []

    matching_parentheses = {
        ')': '(',
        '}': '{',
        ']': '['
    }

    for char in input_string:
        if char in "({[":
            stack.append(char)

        elif char in ")}]":
            if not stack or stack[-1] != matching_parentheses[char]:
                return False

            stack.pop()

    return len(stack) == 0

print(q5_parentheses("({[]})()"))  # Should return True
print(q5_parentheses("({[}])"))    # Should return False
````

Write a Python program to to check if parentheses are balanced using a queue. Use the method `q7_balanced`, which returns `True` if balanced. Otherwise, `False`. Parentheses are balanced when for every opening parenthesis, there is an appropriately placed closing parenthesis. Hence:
```python
    ({[]})()
```
is balanced, while:
```python
    ({[}])
```
is not.

````python
from collections import deque

def q7_balanced(input_string):
    """Check if the parentheses in the input string are balanced using a queue."""
    # Initialize an empty queue
    queue = deque()

    matching_parentheses = {
        ')': '(',
        '}': '{',
        ']': '['
    }

    for char in input_string:
        if char in "({[":
            queue.append(char)

        elif char in ")}]":
            if not queue or queue[-1] != matching_parentheses[char]:
                return False

            queue.pop()

    return len(queue) == 0

print(q7_balanced("({[]})()"))  # Should return True
print(q7_balanced("({[}])"))    # Should return False
````

#### Question 8&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;(10 Points) 
Using the `queue` class from the `Queue` module, write a function `q8_sortqueue` to sort a queue of integers without using any other interim data structure. The function should return the sorted queue as a list. The queue is essentially to be sorted "in-place".

Create a queue with the elements 30, 11, 15 and assign it to the variable `q8`. Use `q8_sortqueue(q8)` to check if the method is correctly working.

````python
from queue import Queue

def min_index(q, sorted_index):
    """Find the index of the smallest element in the queue up to the sorted_index."""
    min_index = -1
    min_value = float('inf')
    n = q.qsize()

    for i in range(n):
        current = q.get()
        if current <= min_value and i <= sorted_index:
            min_index = i
            min_value = current
        q.put(current)

    return min_index
def insert_min_to_rear(q, min_index):
    """Rotate the queue to move the element at min_index to the rear."""
    min_value = None
    n = q.qsize()

    for i in range(n):
        current = q.get()

        if i != min_index:
            q.put(current)
        else:
            min_value = current

    q.put(min_value)

def q8_sortqueue(q):
    """Sort the queue using no extra data structure."""
    n = q.qsize()

    for i in range(n):
        min_idx = min_index(q, n - i - 1)
        insert_min_to_rear(q, min_idx)

    return list(q.queue)

q8 = Queue()
q8.put(30)
q8.put(11)
q8.put(15)

sorted_q8 = q8_sortqueue(q8)
print("Sorted Queue:", sorted_q8)  # Should print [11, 15, 30]
````

Create a queue with the elements 1, 2, 3, 4, 5 and assign it to the variable `q9`. Use `q9_reversequeue(q9)` to check if the method is correctly working.

````python
from queue import Queue

class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        """Add an item to the stack."""
        self.items.append(item)

    def pop(self):
        """Remove and return the top item from the stack."""
        if self.is_empty():
            raise IndexError("pop from an empty stack")
        return self.items.pop()

    def is_empty(self):
        """Check if the stack is empty."""
        return len(self.items) == 0

def q9_reversequeue(q):
    """Reverse the elements in a queue and return them as a list."""
    stack = Stack()

    while not q.empty():
        stack.push(q.get())

    while not stack.is_empty():
        q.put(stack.pop())

    return list(q.queue)
q9 = Queue()
for i in range(1, 6):
    q9.put(i)
reversed_q9 = q9_reversequeue(q9)
print("Reversed Queue:", reversed_q9)  # Should print [5, 4, 3, 2, 1]
````

#### Question 10&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;(10 Points)
Write a Python function, `q10_is_palindrome`, to check if a sequence of characters is a [palindrome](https://en.wikipedia.org/wiki/Palindrome) by using a Stack **and** a Queue. In comparing characters, all whitespace and the case of the characters should be ignored - however, puncatuation should be considered. For example:
```python
   "Live on time, emit no evil"
```
is a palindrome, while:
```python
    "Was it a car or a cat I saw?"
```
is not because of the unbalanced `'?'` at the end.

If the input string is a palindrome, return `True`. Otherwise, `False`.

````python
from queue import Queue

class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        """Add an item to the stack."""
        self.items.append(item)

    def pop(self):
        """Remove and return the top item from the stack."""
        if self.is_empty():
            raise IndexError("pop from an empty stack")
        return self.items.pop()

    def is_empty(self):
        """Check if the stack is empty."""
        return len(self.items) == 0

    def __str__(self):
        """Return a string representation of the stack."""
        return "Stack: " + str(self.items)


def q10_is_palindrome(s):
    """Check if a sequence of characters is a palindrome using a stack and a queue."""
    stack = Stack()
    queue = Queue()

    cleaned_string = ''.join(c for c in s if not c.isspace()).lower()

    for char in cleaned_string:
        stack.push(char)
        queue.put(char)

    while not stack.is_empty():
        if stack.pop() != queue.get():
            return False

    return True

print(q10_is_palindrome("Live on time, emit no evil"))  # Should return True
print(q10_is_palindrome("Was it a car or a cat I saw?"))  # Should return False
````
